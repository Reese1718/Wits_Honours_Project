```python
import shutil
import site
import subprocess
import sys
from pathlib import Path

subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "uninstall",
        "-y",
        "tensorboard",
        "tensorboard-data-server",
    ],
    check=False,
)

# Remove files left behind by the broken installation.
for site_packages in site.getsitepackages():
    site_packages = Path(site_packages)

    for pattern in (
        "tensorboard",
        "tensorboard-*.dist-info",
        "tensorboard_data_server*",
    ):
        for path in site_packages.glob(pattern):
            print("Removing:", path)

            if path.is_dir():
                shutil.rmtree(path)
            else:
                path.unlink()

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "--no-cache-dir",
    "tensorboard",
])

print("TensorBoard reinstalled. Restart the session now.")
```

```python
from pathlib import Path

training_dir = Path(
    "/content/drive/MyDrive/Wits_Honours_Project/pipeline/training"
)

files = [
    training_dir / "train_sasrec.py",
    training_dir / "train_bert4rec.py",
    training_dir / "train_gru4rec.py",
]

compatibility_code = """import numpy as np

# RecBole compatibility with NumPy 2.x
if not hasattr(np, "float_"):
    np.float_ = np.float64

"""

for path in files:
    if not path.exists():
        print("Skipped; file not found:", path)
        continue

    text = path.read_text(encoding="utf-8")

    if "np.float_ = np.float64" in text:
        print("Already patched:", path.name)
        continue

    recbole_position = text.find("from recbole")

    if recbole_position == -1:
        raise RuntimeError(f"No RecBole import found in {path}")

    text = (
        text[:recbole_position]
        + compatibility_code
        + text[recbole_position:]
    )

    path.write_text(text, encoding="utf-8")
    print("Patched:", path.name)
```
