```python
from pathlib import Path

path = Path(
    "pipeline/configs/video_games_augmentation/bert4rec_video_games_llama.yaml"
)

lines = path.read_text().splitlines()

# Find the evaluation section boundaries
eval_start = next(
    i for i, line in enumerate(lines)
    if line.strip() == "eval_args:"
)

hardware_start = next(
    i for i, line in enumerate(lines[eval_start:], start=eval_start)
    if line.strip().startswith("# Hardware")
)

# Keep comments immediately preceding Hardware intact
cut_end = hardware_start
while cut_end > eval_start and lines[cut_end - 1].strip().startswith("#"):
    cut_end -= 1

new_block = [
    "eval_args:",
    "  group_by: user",
    "  order: TO",
    "  split:",
    "    LS: valid_and_test",
    "  mode: full",
    "",
    "metrics:",
    "  - Recall",
    "  - MRR",
    "  - NDCG",
    "  - Hit",
    "",
    "topk: 10",
    "valid_metric: NDCG@10",
    "metric_decimal_place: 4",
    "",
]

lines = lines[:eval_start] + new_block + lines[cut_end:]

path.write_text("\n".join(lines) + "\n")

print("Rebuilt LLaMA evaluation block.")
```

```python
import yaml

path = "pipeline/configs/video_games_augmentation/bert4rec_video_games_llama.yaml"

with open(path) as f:
    config = yaml.safe_load(f)

print("YAML parsed successfully.")
print("Dataset:", config["dataset"])
print("Train batch:", config["train_batch_size"])
print("Eval batch:", config["eval_batch_size"])
print("Learning rate:", config["learning_rate"])
print("Metrics:", config["metrics"])
print("Eval args:", config["eval_args"])
```
